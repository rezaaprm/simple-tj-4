### PENGAMBILAN KOORDINAT DARI DATA .SHP UNTUK DATA POI

##### Title : QGIS Lesson 11: Extracting points to XY coordinates | tutorial video
##### Link : https://youtu.be/y8JKVciv26g?si=Z3-FE3VV0_kGRJ_a

Berikut adalah langkah-langkah detail dan terurut untuk mengekstrak titik koordinat dari file format Shapefile (`.shp`) di QGIS hingga mengekspornya kembali, sesuai dengan tutorial dalam video tersebut:

### **Langkah 1: Mengimpor Data Layer Titik (.shp)**

1. Buka aplikasi **QGIS** [00:05](http://www.youtube.com/watch?v=y8JKVciv26g&t=5).
2. Klik menu **Layer** di bagian atas, pilih **Add Layer**, lalu klik **Add Vector Layer...** [00:06](http://www.youtube.com/watch?v=y8JKVciv26g&t=6).
3. Pada jendela Data Source Manager yang terbuka, pastikan tipe sumber data terpilih sebagai **File** [00:08](http://www.youtube.com/watch?v=y8JKVciv26g&t=8).
4. Klik tombol **Browse** (ikon tiga titik `...`) di samping kanan kotak *Vector Dataset(s)* [00:08](http://www.youtube.com/watch?v=y8JKVciv26g&t=8).
5. Cari folder penyimpanan Anda, pilih file berformat **.shp** (dalam video dicontohkan file `BusTerminals_Mindanao.shp`), lalu klik **Open** [00:09](http://www.youtube.com/watch?v=y8JKVciv26g&t=9).
6. Klik tombol **Add** lalu klik **Close** [00:11](http://www.youtube.com/watch?v=y8JKVciv26g&t=11). Data titik Anda kini sudah muncul di peta kerja [00:13](http://www.youtube.com/watch?v=y8JKVciv26g&t=13).

---

### **Langkah 2: Menambahkan Basemap (Opsional)**

*Langkah ini ditunjukkan dalam video agar visualisasi titik di peta terlihat lebih jelas dan bermakna [00:15](http://www.youtube.com/watch?v=y8JKVciv26g&t=15).*

1. Pada panel **Browser** di sebelah kiri, cari opsi **XYZ Tiles** [00:17](http://www.youtube.com/watch?v=y8JKVciv26g&t=17).
2. Klik kanan pada basemap yang Anda inginkan (misalnya **Bing Aerial Layer**), lalu pilih **Add Layer to Project** [00:21](http://www.youtube.com/watch?v=y8JKVciv26g&t=21).
3. Geser layer basemap tersebut ke urutan paling bawah di panel *Layers* agar tidak menutupi data titik Anda [00:25](http://www.youtube.com/watch?v=y8JKVciv26g&t=25).
4. *(Opsional)* Jika warna titik kurang terlihat, Anda bisa klik kanan pada layer titik > **Properties** > **Symbology** untuk mengubah ukuran atau warna titik menjadi lebih mencolok (misalnya warna kuning) [00:30](http://www.youtube.com/watch?v=y8JKVciv26g&t=30).

---

### **Langkah 3: Memeriksa Atribut Data Sebelum Diekstrak**

1. Klik kanan pada layer titik Anda di panel *Layers*, lalu pilih **Open Attribute Table** [00:51](http://www.youtube.com/watch?v=y8JKVciv26g&t=51).
2. Anda akan melihat bahwa pada tabel atribut bawaan file `.shp` tersebut, **belum terdapat kolom koordinat X dan Y** [00:54](http://www.youtube.com/watch?v=y8JKVciv26g&t=54). Tutup kembali tabel tersebut [01:02](http://www.youtube.com/watch?v=y8JKVciv26g&t=62).

---

### **Langkah 4: Mengekstrak Titik Menjadi Koordinat X & Y**

1. Klik kanan kembali pada layer titik Anda, pilih **Export**, lalu klik **Save Features As...** [01:08](http://www.youtube.com/watch?v=y8JKVciv26g&t=68).
2. Pada opsi **Format**, ubah dari *ESRI Shapefile* menjadi **Comma Separated Value [CSV]** [01:13](http://www.youtube.com/watch?v=y8JKVciv26g&t=73).
3. Pada bagian **File name**, klik tombol tiga titik `...` untuk menentukan lokasi penyimpanan file, beri nama file baru Anda (misalnya `BusTerminals_Coordinates`), lalu klik **Save** [01:21](http://www.youtube.com/watch?v=y8JKVciv26g&t=81).
4. Periksa bagian **CRS** (Coordinate Reference System), pastikan menggunakan sistem proyeksi yang sesuai (dalam video menggunakan **EPSG:4326 - WGS 84**) [01:41](http://www.youtube.com/watch?v=y8JKVciv26g&t=101).
5. Gulir ke bawah hingga Anda menemukan bagian **Layer Options** [01:44](http://www.youtube.com/watch?v=y8JKVciv26g&t=104).
6. Di dalam kotak parameter **GEOMETRY**, ketik perintah berikut dengan huruf kapital:
```text
AS_XY

```


*(Catatan: Langkah ini adalah kunci utama untuk memerintahkan QGIS memunculkan kolom koordinat X dan Y pada file CSV hasil ekspor).* [02:08](http://www.youtube.com/watch?v=y8JKVciv26g&t=128)
7. Pastikan centang opsi **"Add saved file to map"** di bagian bawah tetap aktif, lalu klik **OK** [02:18](http://www.youtube.com/watch?v=y8JKVciv26g&t=138).

---

### **Langkah 5: Memverifikasi Hasil Ekspor**

1. Layer baru berformat `.csv` otomatis akan masuk ke dalam panel *Layers* QGIS Anda [02:22](http://www.youtube.com/watch?v=y8JKVciv26g&t=142).
2. Klik kanan pada layer CSV baru tersebut, lalu pilih **Open Attribute Table** [02:27](http://www.youtube.com/watch?v=y8JKVciv26g&t=147).
3. Sekarang, Anda akan melihat dua kolom baru di sisi paling kiri tabel, yaitu kolom **X** (Bujur/Longitude) dan kolom **Y** (Lintang/Latitude) yang berisi titik koordinat dari masing-masing objek [02:29](http://www.youtube.com/watch?v=y8JKVciv26g&t=149).
4. Anda juga bisa membuka file `.csv` yang telah tersimpan tadi langsung menggunakan aplikasi **Microsoft Excel** untuk keperluan pengolahan data lainnya [02:46](http://www.youtube.com/watch?v=y8JKVciv26g&t=166).

````python
import os
import pandas as pd
import warnings

warnings.filterwarnings("ignore")

# Definisikan path folder dataset
dataset_path = r"D:\UPNVJ\X_Proposal\Dataset\gtfs-test\transjakarta"

# Memperbarui masa berlaku kalender
calendar_file = os.path.join(dataset_path, "calendar.txt")
df_calendar = pd.read_csv(calendar_file)

df_calendar.loc[df_calendar['service_id'] == 'X', 'end_date'] = 20271231
df_calendar.to_csv(calendar_file, index=False)
print("Aksi 1 selesai: Masa berlaku kalender diperbarui.")
````

````python
# Memperbarui dan melengkapi dokumen feed info
feedinfo_file = os.path.join(dataset_path, "feed_info.txt")
df_feed = pd.DataFrame([{
    "feed_publisher_name": "Transjakarta",
    "feed_publisher_url": "https://transjakarta.co.id/",
    "feed_lang": "id",
    "feed_contact_email": "naektije@transjakarta.co.id",
    "feed_contact_url": "https://transjakarta.co.id/",
    "feed_start_date": 20040115,
    "feed_end_date": 20271231,
    "feed_version": "2.0.0"
}])
df_feed.to_csv(feedinfo_file, index=False, lineterminator='\r\n')
print("Aksi 2 selesai: File feed_info.txt berhasil diperkaya.")
````

````python
# Mengubah format penulisan nama halte agar tidak kapital murni
stops_file = os.path.join(dataset_path, "stops.txt")
df_stops = pd.read_csv(stops_file)

def to_mixed_case(name):
    if pd.isna(name):
        return name
    title_named = str(name).title()
    title_named = title_named.replace("Rs ", "RS ").replace("Lrt ", "LRT ").replace("Dpr ", "DPR ")
    return title_named

df_stops['stop_name'] = df_stops['stop_name'].apply(to_mixed_case)
df_stops.to_csv(stops_file, index=False)
print("Aksi 3 selesai: Format penulisan nama halte distandarisasi.")
````

````python
# Mengunci dokumen kalender pengecualian ke dalam struktur standar internasional
calendardates_file = os.path.join(dataset_path, "calendar_dates.txt")
with open(calendardates_file, 'w', encoding='utf-8', newline='\r\n') as f:
    f.write("service_id,date,exception_type\n")
print("Aksi 4 selesai: Dokumen calendar_dates.txt direset ke format murni.")
````

````python
# Memperbaiki urutan dan nilai jarak pada bentuk rute spasial
shapes_file = os.path.join(dataset_path, "shapes.txt")
df_shapes = pd.read_csv(shapes_file)
df_shapes.columns = df_shapes.columns.str.strip()

# Penyesuaian nilai desimal untuk menghindari warning penumpukan koordinat
eps = 0.001
for shape_id, seq in [('mx84', 71), ('5C-R09_shp', 19)]:
    mask = (df_shapes['shape_id'] == shape_id) & (df_shapes['shape_pt_sequence'] == seq)
    if mask.any():
        prev_val = df_shapes.loc[df_shapes['shape_id'] == shape_id, 'shape_dist_traveled'].shift(1).loc[mask].values
        if len(prev_val) > 0 and pd.notna(prev_val[0]):
            df_shapes.loc[mask, 'shape_dist_traveled'] = prev_val[0] + eps

# Perbaikan titik krusial pada sekuens koridor 13
mask_seq53 = (df_shapes["shape_id"] == "13-R02_shp") & (df_shapes["shape_pt_sequence"] == 53)
if mask_seq53.any():
    df_shapes.loc[mask_seq53, "shape_dist_traveled"] = 3689.153250

df_shapes = df_shapes.sort_values(['shape_id', 'shape_pt_sequence']).reset_index(drop=True)
df_shapes.to_csv(shapes_file, index=False, float_format="%.6f", lineterminator='\r\n')
print("Aksi 5 selesai: Sinkronisasi jarak geometris jalur busway berhasil dikunci.")
````

````python
# Melakukan validasi tipe data integer dan membatasi jarak trip agar tidak melebihi bentuk rute
stoptimes_file = os.path.join(dataset_path, "stop_times.txt")
df_stoptimes = pd.read_csv(stoptimes_file)
df_stoptimes.columns = df_stoptimes.columns.str.strip()

max_shape_dist = df_shapes.groupby("shape_id")["shape_dist_traveled"].max().to_dict()

trips_file = os.path.join(dataset_path, "trips.txt")
df_trips = pd.read_csv(trips_file)
df_trips.columns = df_trips.columns.str.strip()
trip_to_shape = dict(zip(df_trips['trip_id'], df_trips['shape_id']))

df_stoptimes['shape_dist_traveled'] = df_stoptimes['shape_dist_traveled'].astype(float)

# Mengoreksi baris halte paling terakhir pada setiap perjalanan secara presisi
last_idx_per_trip = df_stoptimes.groupby('trip_id')['stop_sequence'].idxmax()
for idx in last_idx_per_trip:
    trip_id = df_stoptimes.loc[idx, 'trip_id']
    shape_id = trip_to_shape.get(trip_id)
    if shape_id and shape_id in max_shape_dist:
        max_allowed = max_shape_dist[shape_id]
        if df_stoptimes.loc[idx, 'shape_dist_traveled'] > max_allowed:
            df_stoptimes.loc[idx, 'shape_dist_traveled'] = max_allowed

# Bersihkan data kosong (NaN) terlebih dahulu sebelum diubah ke integer
df_stoptimes['stop_sequence'] = df_stoptimes['stop_sequence'].fillna(0).astype('int64')
df_stoptimes['timepoint'] = df_stoptimes['timepoint'].fillna(0).astype('int64')

if 'pickup_type' in df_stoptimes.columns:
    df_stoptimes['pickup_type'] = df_stoptimes['pickup_type'].fillna(0).astype('int64')
if 'drop_off_type' in df_stoptimes.columns:
    df_stoptimes['drop_off_type'] = df_stoptimes['drop_off_type'].fillna(0).astype('int64')

# Urutkan stop_times berdasarkan trip_id dan stop_sequence
df_stoptimes = df_stoptimes.sort_values(by=['trip_id', 'stop_sequence']).reset_index(drop=True)

df_stoptimes.to_csv(stoptimes_file, index=False, float_format='%.6f', lineterminator='\r\n')
print("Aksi 6 selesai: Sinkronisasi dokumen stop_times.txt berhasil diamankan tanpa error data kosong.")
````

````python
# Membersihkan data halte kosong yang tidak terikat pada jadwal operasional, transfer, atau stasiun utama
transfers_file = os.path.join(dataset_path, "transfers.txt")

used_stops = set(df_stoptimes['stop_id'].unique())

if os.path.exists(transfers_file):
    df_transfers = pd.read_csv(transfers_file)
    used_in_transfers = set(df_transfers['from_stop_id'].unique()) | set(df_transfers['to_stop_id'].unique())
    used_stops.update(used_in_transfers)

parent_stations = set(df_stops[df_stops['location_type'] == 1]['stop_id'])
used_stops.update(parent_stations)

df_stops_filtered = df_stops[df_stops['stop_id'].isin(used_stops)]
df_stops_filtered.to_csv(stops_file, index=False, lineterminator='\r\n')
print(f"Aksi 7 selesai: Penyaringan halte aktif berhasil. Jumlah halte diperbarui.")
print("Seluruh rangkaian pembersihan dataset selesai dilaksanakan.")
````
